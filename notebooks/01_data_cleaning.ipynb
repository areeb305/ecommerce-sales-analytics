# Data Cleaning — Online Retail Dataset

## Objective

The purpose of this notebook is to inspect the cleaned Online Retail dataset and verify that the data-cleaning pipeline produced an analysis-ready dataset.

The raw dataset contains transaction-level records including invoice information, products, quantities, prices, customers, dates, and countries.

The cleaning process addresses:

- Duplicate transactions
- Cancelled invoices
- Non-positive quantities
- Non-positive unit prices
- Missing product descriptions
- Invalid invoice dates

Transactions with missing Customer IDs are retained because they can still contribute to overall sales, product, country, and time-based analyses. Customer-specific analyses will use only transactions with valid Customer IDs.

```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/cleaned_sales_data.csv",
    parse_dates=["invoicedate"]
)

print("Dataset loaded successfully.")
```

```python
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
```

```python
print("Duplicate rows:", df.duplicated().sum())

print(
    "Cancelled invoices:",
    df["invoiceno"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

print(
    "Non-positive quantities:",
    (df["quantity"] <= 0).sum()
)

print(
    "Non-positive prices:",
    (df["unitprice"] <= 0).sum()
)
```

## Cleaning Summary

The raw dataset contained **541,909 rows and 8 columns**.

After cleaning, the analysis-ready dataset contains **524,878 rows and 13 columns**.

A total of **17,031 rows** were removed during the cleaning process, representing approximately **3.14% of the original dataset**.

### Cleaning Decisions

The following steps were applied:

1. Removed exact duplicate transactions.
2. Converted invoice dates to a valid datetime format.
3. Removed cancelled invoices.
4. Removed transactions with non-positive quantities.
5. Removed transactions with zero or negative unit prices.
6. Removed records without product descriptions.
7. Retained transactions with missing Customer IDs for non-customer-specific analysis.
8. Created a `sales` variable using quantity × unit price.
9. Created year, month, month name, and day-of-week variables for time-based analysis.

The resulting dataset is suitable for exploratory sales analysis while preserving anonymous transactions that still contain useful business information.

```python
raw_rows = 541909
clean_rows = len(df)

removed_rows = raw_rows - clean_rows
removed_percentage = (removed_rows / raw_rows) * 100

print(f"Raw rows: {raw_rows:,}")
print(f"Clean rows: {clean_rows:,}")
print(f"Rows removed: {removed_rows:,}")
print(f"Percentage removed: {removed_percentage:.2f}%")
```
